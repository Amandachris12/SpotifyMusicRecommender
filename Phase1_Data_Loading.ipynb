```python
import pandas as pd
```

```python
# 1. Read the main dataset
data = pd.read_csv("data (1).csv")
# 2. Read the genre dataset
genre_data = pd.read_csv("data_by_genres.csv")

# 3. Read the year dataset
year_data = pd.read_csv("data_by_year.csv")

# 4. Read the artist dataset
artist_data = pd.read_csv("data_by_artist.csv")
```

```python
# 5. Display the first two rows of each dataset
print("Main Dataset:")
print(data.head(2))

print("\nGenre Dataset:")
print(genre_data.head(2))

print("\nYear Dataset:")
print(year_data.head(2))

print("\nArtist Dataset:")
print(artist_data.head(2))
```

```python
# 6. Retrieve information about the main and genre datasets
print("\nMain Dataset Information:")
data.info()

print("\nGenre Dataset Information:")
genre_data.info()
```

```python
# 7. Create a decade column from the year column
data["decade"] = data["year"].apply(lambda x: (x // 10) * 10)

# Display the first few rows to confirm
print("\nDataset with Decade Column:")
print(data[["year", "decade"]].head())
```
